## Dataset 1: Pakistani E-Commerce Data

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\zahid\Desktop\ICA\Assignment 14\pakistani_ecommerce_dataset.csv") 
df.head()
```

```python
df.info()
```

```python
df.info
```

```python
category_count = df.groupby("product_category")["user_id"].count()

print(category_count)
```

Home Decor is the most popular product category, with 30 orders in the dataset.

```python
df.describe()
```

```python
df["order_value"].describe()
```

The average order value in the Pakistani E-Commerce dataset is Rs. 2,832.15.

```python
city_orders = df.groupby("city")["city"].count()

print(city_orders)
```

Karachi has the highest number of orders, with 24 orders in the Pakistani E-Commerce dataset.

```python
payment_count = df.groupby("payment_method")["payment_method"].count()

print(payment_count)
```

Cash on Delivery is the most used payment method, accounting for 38 orders in the Pakistani E-Commerce dataset.

```python
city_average = df.groupby("city")["order_value"].mean()

print(city_average)
```

```python
plt.bar(city_average.index, city_average.values)

plt.xlabel("City")
plt.ylabel("Average Order Value")
plt.title("Average Order Value by City")

plt.show()
```

The bar chart compares the average order value across different cities. Islamabad has the highest average order value, whereas Rawalpindi has the lowest.

## Data Set 2: Student Lifestyle & Study Habits 

```python
students = pd.read_csv(r"C:\Users\zahid\Desktop\ICA\Assignment 14\student_lifestyle_dataset.csv")
```

```python
students["study_hours"].describe()
```

The average number of study hours among the students is approximately 4.40 hours per day.

```python
sleep_average = students.groupby("gender")["sleep_hours"].mean()

print(sleep_average)
```

Female students sleep more on average than male students in this dataset.

```python
correlation = students["screen_time"].corr(students["sleep_hours"])
print("Correlation:", correlation)
```

The correlation is 0.052, which is very close to zero. This means there is almost no linear relationship between screen time and sleep hours in this dataset.

```python
plt.hist(students["daily_transport_cost"], bins=10,
         edgecolor='black')

plt.title("Daily Transport Cost")
plt.xlabel("Daily Transport Cost (Rs.)")
plt.ylabel("Number of Students")

plt.show()
```

```python
internet_average = students.groupby("gender")["internet_usage"].mean()

print(internet_average)
```

Male students have higher average internet usage (6.00 hours) than female students (5.32 hours) in this dataset.
