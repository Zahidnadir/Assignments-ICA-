```python
import pandas as pd

df = pd.read_csv(r"C:\Users\zahid\Downloads\teaching_prices_clean.csv")

df["date"] = pd.to_datetime(df["date"])

display(df.head())

print("Dataset shape:", df.shape)
```

```python
print("Items:", sorted(df["item"].unique()))
print("Cities:", sorted(df["city"].unique()))
print("Number of observation dates:", df["date"].nunique())


print("\nMissing values:")
print(df[["date", "item", "city", "price"]].isnull().sum())
```

```python
averages = (
    df.groupby(["item", "city"])["price"]
      .agg(mean="mean", median="median")
      .reset_index()
)

averages[["mean", "median"]] = averages[["mean", "median"]].round(2)

display(averages)
```

```python
df_sorted = df.sort_values("date")

first_last = (
    df_sorted.groupby(["item", "city"])["price"]
    .agg(first_price="first", last_price="last")
    .reset_index()
)

first_last["percentage_change"] = (
    (first_last["last_price"] - first_last["first_price"])
    / first_last["first_price"]
) * 100

first_last[["first_price", "last_price", "percentage_change"]] = (
    first_last[["first_price", "last_price", "percentage_change"]].round(2)
)

display(first_last)
```

```python
results = averages.merge(
    first_last,
    on=["item", "city"]
)

results = results[
    ["item", "city", "mean", "median",
     "first_price", "last_price", "percentage_change"]
]

display(results)
```

Brief interpretation

Use the results table above to support the following points:

1. Compare the mean and median for each item and city.
2. Identify which item had the largest first-to-last percentage increase.
3. Identify which item had the smallest first-to-last percentage increase.
4. Remember that these calculations describe the supplied observations. They do not, by themselves, explain what caused prices to change.
