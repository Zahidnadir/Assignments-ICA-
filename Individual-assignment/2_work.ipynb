```python
print("Dataset Shape:", df.shape)

print("Distinct Items:", df['item'].nunique())

print("Distinct Dates:", df['date'].nunique())

print("Distinct Cities:", df['city'].nunique())

print("\nColumns:")
print(df.columns.tolist())
```

```python
comparison = df.pivot_table(
    index=['date', 'item'],
    columns='city',
    values='price'
)

comparison['City 2 Higher'] = (
    comparison['City 2'] > comparison['City 1']
)

print(comparison)

print("\nIs Claim 1 supported?")
print(comparison['City 2 Higher'].all())
```

```python
for city in df['city'].unique():

    city_data = df[df['city'] == city]

    average_prices = city_data.groupby('item')['price'].mean()

    print("\n", city)
    print(average_prices.sort_values(ascending=False))
```
