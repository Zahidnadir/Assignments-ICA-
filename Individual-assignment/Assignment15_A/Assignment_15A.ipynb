```python
import matplotlib.pyplot as plt
```

```python
average_fare = df.groupby("service_type")["fare"].mean()
print(average_fare)

plt.bar(average_fare.index, average_fare.values, color="green", width=0.5)

plt.xlabel("Service Type")
plt.ylabel("Average Fare (Rs.)")
plt.title("Average Fare by Service Type")

plt.show()
```
