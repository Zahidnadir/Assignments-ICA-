```python
import pandas as pd

df = pd.read_csv(r"C:\Users\zahid\Downloads\teaching_prices_messy.csv")

print(df.head())
print("Shape:", df.shape)
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("Duplicate Records:", df.duplicated().sum())

print("Product Names:")
print(df["item"].unique())

print("Cities:")
print(df["city"].unique())
```

```python
clean = df.copy()

clean = clean.drop_duplicates()

clean["price"] = pd.to_numeric(
    clean["price"].astype(str).str.replace(",", ""),
    errors="coerce"
)

print(clean.head())

print("Rows Before Cleaning:", len(df))
print("Rows After Cleaning:", len(clean))
```

```python
print("Missing Values:")
print(clean.isnull().sum())

print("Remaining Duplicates:", clean.duplicated().sum())

print("Unusual Prices:")
print(clean[clean["price"] > 1000])

clean.to_csv("teaching_prices_audited.csv", index=False)

print("Cleaned dataset saved successfully.")
```

Conclusion
The dataset contained duplicate records, missing prices, unusual values, and inconsistent product names. I removed exact duplicates and converted prices into numeric format. Other questionable values were retained for further investigation to avoid making changes without proper evidence.
